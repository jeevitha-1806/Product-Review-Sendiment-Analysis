```python
df = df.drop_duplicates()
print(df)
```

```python
df = df.dropna()
print(df)
```

```python
df = df.reset_index(drop=True)
print(df)
```

```python
df["REVIEW"] = df["REVIEW"].str.lower()
print(df["REVIEW"])
```

```python
df["REVIEW"] = df["REVIEW"].str.strip()
print(df["REVIEW"])
```

```python
print(df.head())
print(df.shape)
print(df.isnull().sum())
```

```python
df["Review_Length"] = df["REVIEW"].str.len()
print(df[["REVIEW", "Review_Length"]])
```

```python
print("Mean:", df["Review_Length"].mean())
```

```python
print("Median:", df["Review_Length"].median())
```

```python
print("Mode:", df["Review_Length"].mode()[0])
```

```python
print("Minimum:", df["Review_Length"].min())
```

```python
print("Maximum:", df["Review_Length"].max())
```

```python
range_value = df["Review_Length"].max() - df["Review_Length"].min()
print("Range:", range_value)
```

```python
print("Variance:", df["Review_Length"].var())
```

```python
print("Standard Deviation:", df["Review_Length"].std())
```

```python
print(df["SENTIMENT"].value_counts())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Review_Length"], bins=10)
plt.xlabel("Review Length")
plt.ylabel("Frequency")
plt.title("Distribution of Review Length")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["Review_Length"])
plt.ylabel("Review Length")
plt.title("Box Plot of Review Length")
plt.show()
```

```python
sentiment_counts = df["SENTIMENT"].value_counts()
plt.figure(figsize=(8,5))
plt.bar(sentiment_counts.index, sentiment_counts.values)
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.title("Sentiment Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(range(len(df)), df["Review_Length"])
plt.xlabel("Review Index")
plt.ylabel("Review Length")
plt.title("Review Length Scatter Plot")
plt.show()
```
